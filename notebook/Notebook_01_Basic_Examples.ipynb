## Esempio 10 – Grafico di una distribuzione gaussiana
Questo programma genera numeri casuali con distribuzione normale e li visualizza con un grafico.

**Attenzione**: durante l'esecuzione si otterranno errori per la mancanza dei moduli `numpy` (libreria matematica), `matplotlib` (libreria per grafici) o `scipy` (libreria statistica), questo è voluto per mostrare l'utilizzo di pip. Installare le due librerie con pip e riprovare.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Genera 1000 valori con distribuzione normale (media=0, deviazione=1)
dati = np.random.normal(loc=0, scale=1, size=1000)

# Stima dei parametri della gaussiana (fit)
mu, sigma = norm.fit(dati)

# Genera la curva della gaussiana stimata
x = np.linspace(min(dati), max(dati), 200)
gaussiana = norm.pdf(x, mu, sigma)

plt.figure(figsize=(8, 5))

# Istogramma dei dati
plt.hist(dati, bins=30, density=True, color='skyblue', edgecolor='black', alpha=0.6)

# Curva della gaussiana stimata
plt.plot(x, gaussiana, 'r-', linewidth=2, label=f"Fit Gaussiana (mu={mu:.2f}, sigma={sigma:.2f})")

plt.title("Distribuzione Gaussiana con Fit")
plt.xlabel("Valori")
plt.ylabel("Densità")
plt.grid(True)
plt.legend()
plt.show()
```
