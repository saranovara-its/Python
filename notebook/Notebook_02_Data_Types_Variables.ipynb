Questo notebook introduce alcuni concetti fondamentali del linguaggio Python tramite esempi:

- Conversioni di tipo e funzioni built-in
- Variabili e scope
- Operatori aritmetici, logici e di confronto

L'obiettivo non è una discussione completa di questi concetti, ma mostrare attraverso esempi semplici alcune delle caratteristiche peculiari del linguaggio Python.

## Numeri interi (int)

Gli interi rappresentano numeri senza decimali.  
Python può gestire numeri molto grandi senza generare overflow. Non esiste un MAX INT, gli interi crescono fino a che c'è memoria disponibile. Quando un numero cresce oltre la capacità di un normale intero, Python passa automaticamente a una rappresentazione più ampia.

Python ha un solo tipo di intero (`int`), non esistono `short`, `long`, etc. (a dimensione fissa).


````python
x = 10
y = 10**100  # numero enorme
print(type(x), x)
print(type(y), y)
````

````python
# Confronto con C

x = 2147483647 # max int 32-bit (2^31 - 1), 1 bit per il segno
x = x + 1
print(x)  # 2147483648
````

## Numeri decimali (float)

I float rappresentano numeri con la virgola.  
La loro rappresentazione interna è binaria, quindi alcuni valori non possono essere memorizzati con precisione assoluta.  
Questo porta a risultati che possono sembrare sbagliati quando si fanno operazioni semplici.

````python
a = 0.1
b = 0.2
c = a + b

print(c)
print(0.1 + 0.2)
print(0.3)
print(0.3 == 0.1 + 0.2)
````

````python
from decimal import Decimal

a = Decimal("0.1")
b = Decimal("0.2")
print(a + b)
````

## Stringhe (str)

Le stringhe sono sequenze di caratteri.  
Si possono creare con:
- apici singoli ' '
- apici doppi " "
- triple """ """ per testi multilinea

Le stringhe non possono essere modificate dopo la creazione.  
L’operatore `+` concatena, mentre `*` ripete.


````python
s = "ciao"
print(s * 3)
# s[0] = "C"  # errore: le stringhe non sono modificabili
````

## Booleani (bool)

I booleani rappresentano valori logici: `True` e `False`.

In Python alcuni valori vengono interpretati come falsi:
- 0
- ""
- []
- None

`True` e `False` sono anche numeri: `True` vale 1 e `False` vale 0.

````python
print(True + True)
print(False + 10)
print(True + 10)
print(bool(0), bool("ciao"))
````

## Conversioni esplicite: int(), float(), str(), bool()

Python permette di convertire i tipi manualmente.

Esempi utili:
- `int("10")` funziona
- `int("10a")` genera errore
- `bool("False")` restituisce True perché la stringa non è vuota


````python
print(int("10"))
# print(int("10a"))  # ValueError
print(bool("False"))
````

## Funzione type()

`type()` restituisce il tipo di un valore o variabile.


````python
print(type(10))
print(type(10.5))
print(type("ciao"))
print(type(True))
````

## Funzione isinstance()

`isinstance()` verifica se un oggetto appartiene a un certo tipo.  
È utile per controlli più precisi rispetto a `type()`.


````python
print(isinstance(10, int))
print(isinstance("ciao", str))
print(isinstance(True, bool))
````

## Conversioni implicite (coercion)

Durante alcune operazioni Python converte automaticamente i tipi:
- int + float → float
- True + 10 → 11 (perché True vale 1)


````python
print(10 + 2.5)
print(True + 10)
````

## Funzioni built-in utili: len(), round(), abs()

Queste funzioni permettono di lavorare rapidamente con stringhe e numeri.


````python
print(len("ciao"))
print(round(3.14159, 2))
print(abs(-10))
````

## Variabili: assegnazione dinamica

Python non richiede dichiarazioni di tipo.  
Una variabile può cambiare tipo durante l’esecuzione.


````python
x = 10
print(type(x))
x = "ciao"
print(type(x))
````

## Shadowing: una variabile può nascondere un nome esistente

Se si assegna un valore a un nome già usato da una funzione built-in, la funzione non sarà più accessibile.


````python
len = 10  # sovrascrive la funzione len
print(len)
# print(len("ciao"))  # errore: len non è più una funzione
````

## Scope locale e globale

Le variabili definite dentro una funzione esistono solo all’interno della funzione.  
Quelle definite fuori sono globali.


````python
x = "globale"

def test():
    x = "locale"
    print(x)

test()
print(x)
````

## Uso di global

La parola chiave `global` permette a una funzione di modificare una variabile definita fuori dalla funzione.


````python
x = 0

def increment():
    global x
    x += 1

increment()
print(x)
````

## Variabili non inizializzate

Usare una variabile prima di assegnarle un valore genera errore.


````python
# print(a)  # NameError
a = 10
print(a)
````

## Mutabilità

In Python i tipi si dividono in **immutabili** e **mutabili**.  
La mutabilità indica se un oggetto può essere modificato dopo la creazione.

Ad esempio le stringhe sono immutabili, mentre ad esempio le liste sono mutabili.

````python
s = "ciao"
# s[0] = "C"  # errore

lst = [1, 2, 3] # esempio di lista (array)
lst[0] = 99
print(lst)
````

Questi oggetti **non possono essere modificati**: ogni operazione che sembra cambiarli in realtà crea un nuovo oggetto.
- **int**
- **float**
- **bool**
- **str**
- **tuple**
- **frozenset**
- **bytes**

Esempio: una stringa non può essere modificata in-place.
```python
s = "ciao"
s = s.upper()  # viene creato un nuovo oggetto


I tipi mutabili sono oggetti che **possono essere modificati dopo la creazione**.  
Le modifiche avvengono direttamente sull’oggetto esistente, senza creare nuove copie.  
Questo comportamento è importante quando si lavora con alias, funzioni e strutture dati complesse.

Tipi mutabili principali:
- **list**
- **dict**
- **set**
- **bytearray**: versione mutabile di bytes
- **oggetti custom**: istanze di classi definite dall’utente

Esempio: modifica di una lista

```python
lst = [1, 2, 3]
lst[0] = 99      # modifica in-place
lst.append(4)    # aggiunge un elemento
print(lst)       # [99, 2, 3, 4]

Se due variabili puntano allo stesso oggetto mutabile, una modifica tramite una variabile è visibile anche tramite l’altra.

````python
a = [1, 2]
b = a
b.append(3)
print(a)   # [1, 2, 3]
````

## Operatori aritmetici

- Somma, Differenza: + - 
- Moltiplicazione: * 
- Divisione: / // 
- Modulo (parte intera): % 
- Potenza: **

La divisione `/` restituisce sempre un float.  
La divisione `//` restituisce la parte intera.

````python
print(10 / 3)
print(10 // 3)
print(2 ** 10)
````

## Operatori logici

and, or, not

`and` e `or` non restituiscono necessariamente un booleano.  
Restituiscono uno dei valori coinvolti nell’operazione.


````python
print(True and 10)
print(False or "ciao")
print(not 0)
````

In Python gli operatori logici non restituiscono sempre un valore booleano.  
`and` e `or` ritornano **uno dei valori originali**, in base alla loro logica di valutazione.  
Questo permette di scrivere espressioni più compatte e di usare valori come "fallback" o condizioni implicite.

### Operatore `and`
Restituisce:
- il **primo valore falso** trovato, oppure  
- l’**ultimo valore** se tutti sono veri.

Esempi:

```python
10 and 20      # 20
0 and 20       # 0
"" and "ciao"  # ""
```

### Operatore `or`

Restituisce:
- il **primo valore vero**, oppure
- l’**ultimo valore** se tutti sono falsi.

Esempi:

```python
10 or 20       # 10
0 or 20        # 20
"" or "ciao"   # "ciao"
```

### Operatore `not`

E' l’unico che restituisce sempre un booleano.

Esempi:

```python
not 0      # True
not 10     # False
not ""     # True
```

### Motivazione

Questo comportamento permette idiomi Python molto comuni, come:

```python
nome = input() or "Anonimo"
```

Se l’utente non inserisce nulla ("" è False), viene usata la stringa "Anonimo".

```python
x and funzione(x)
```

La funzione viene chiamata solo se x è True.

## Operatori di confronto

==, !=, <, >, <=, >=

Le stringhe vengono confrontate in ordine alfabetico, non numerico.


````python
print(10 < 2)
print("a" < "b")
print("10" < "2")  # confronto alfabetico
````

## Operatore `in`

Verifica se un elemento è contenuto in una stringa o in una lista.

````python
print("a" in "ciao")
print(2 in [1, 2, 3])
````
